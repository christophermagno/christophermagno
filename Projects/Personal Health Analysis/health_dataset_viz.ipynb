## Ordinary Least Square Analysis Notes

### A strong linear regression model has:
1. A meaningful effect (interpretability)
2. Statistical significance (signal vs noise)
3. Explanatory power (R² / adj R²)
4. Reasonable assumptions (diagnostics)
5. Stability & plausibility (not a fluke)

#### Coefficients that make sense (first filter)
Before looking at any statistic, ask:
- “Does this coefficient mean something real?”

What to look for in an OLS analysis:
* Correct direction
* Reasonable magnitude
* Interpretable units

If a coefficient doesn’t pass the common-sense test, stop there.

#### Statistical significance (signal vs noise)
What to check
* p-value of the coefficient
* F-statistic p-value (overall model)

Rules of thumb
* p < 0.05 → statistically significant
* p < 0.01 → strong evidence
* p ≪ 0.001 → overwhelming evidence

Important nuance
Statistical significance:
* ≠ importance
* ≠ causation

#### Explanatory power (how much does it explain?)
R² and Adjusted R²
This tells you:
“How much of the outcome variability does my model explain?”

Interpreting R² (context matters)
R²	Interpretation
* < 0.05	Very weak
* 0.05–0.15	Weak
* 0.15–0.35	Moderate
* 0.35–0.60	Strong
* > 0.60	Very strong
    
#### Confidence intervals (stability of the estimate)
Never stop at p-values.

What to look for
* Narrow CI → stable estimate
* CI crossing 0 → uncertain effect
* CI width relative to coefficient size
* Large |t| value (+15?)
Example:
* 244 ± ~30 calories/hour → stable
* −6 ± ~4 calories/point → noisy

#### Residual diagnostics (is OLS appropriate?)
Separating “usable” from “fragile”.

🔹 Durbin–Watson
* ~2 → no autocorrelation
* <1.5 → positive autocorrelation (common in time series)

🔹 Normality (Omnibus / JB)
* Large n → less critical
* Small n → more important

🔹 Condition number
* < 30 → great
* 30–1,000 → okay
* 1,000 → scale or collinearity issues

#### Sample size & power (often overlooked)

Rough intuition
* Small n → wide CI → unstable estimates
* Large n → small effects become detectable

A “strong” model needs enough data to support the claim.

Your models mostly passed.

#### The mistake to avoid when looking for meaningful OLS analysis
❌ Chasing:
* lowest p-value
* highest R²
* “significance” alone

✅ Prioritize:
* Interpretability
* Effect size
* Stability
* Plausibility

#### A simple checklist you can reuse

When running a regression, ask:
* Does the coefficient make sense?
* Is it statistically significant?
* How much variance does it explain?
* Are CIs tight and stable?
* Do diagnostics look acceptable?
* Is the sample size sufficient?
* Does this variable add insight beyond others?

If the answer is yes to most, it is a strong model.

#### Final takeaway (this is the mindset shift)

A strong statistical model explains a meaningful amount of variance with interpretable, stable coefficients — not just statistically significant ones.

#### Extra notes
T-Value

The t-value measures how far a coefficient is from zero, in units of its own uncertainty.
| t |  Interpretation  |
|-----|---------------|
| < 1 | Pure noise  |
| 1 – 2 | Weak / unstable |
| ≈ 2 | Barely significant |
| 2 – 3 | Statistically real |
| 3 – 5 | Strong |
| 5 – 10 | Very strong |
| > 10 | Overwhelming |

Reasonable assumptions (diagnostics)
Stability & plausibility (not a fluke)
All five matter — not just p-values.

p-value answers “Is there an effect?”
coefficient answers “How big is the effect?”
A model can be:
statistically significant
but practically unimportant

Coefficient = effect size (magnitude).
Standard error = uncertainty.
t-value = strength of evidence.
Rsq/adj Rsq = explanatory significance. Tiny R means it explains almost nothing about the outcome

```python
def is_int_or_float(column):
    return is_numeric_dtype(column) and not pd.api.types.is_datetime64_any_dtype(column)

def is_ols_meaningful(results):
    """
    Given a RegressionResults class, interpret whether the OLS analysis is
    meaningful or just noise.

    :param results: RegressionResults class
    :return: dict
    """

    key_vars = results.params.index

    # Test 1: Test for overall signal using the F pvalue
    f = results.fvalue
    f_p = results.f_pvalue
    test1 = np.isfinite(f_p) and f_p < 0.05

    # Test 2: Test for coefficient realness
    t = results.tvalues.iloc[1]
    p = results.pvalues.iloc[1]
    coef = results.params.iloc[1]
    std_err = results.bse.iloc[1]

    # CI
    ci = results.conf_int(alpha=0.5)  # 95% CI
    lo, hi = ci.loc[key_vars[1], 0], ci.loc[key_vars[1], 1]
    ci_excludes_zero = not (lo <= 0 <= hi)

    test2 = abs(t) >= 2.0 and ci_excludes_zero

    # Test 3: Test explanatory power
    rsq = results.rsquared
    rsq_adj = results.rsquared_adj
    test3 = np.isfinite(rsq_adj) and rsq_adj >= 0.15

    # Store the results
    out = {
        'fvalue': results.fvalue,
        'f_pvalue': results.f_pvalue,
        'rsquared': results.rsquared,
        'rsquared_adj': results.rsquared_adj,
        'coeff_const': results.params.iloc[0],
        f'coeff_target': results.params.iloc[1],
        f'std_err_const': results.bse.iloc[1],
        f'std_err_target': results.bse.iloc[1],
        'pvalue_const': results.pvalues.iloc[0],
        f'pvalue_target': results.pvalues.iloc[1],
        'tvalue_const': results.tvalues.iloc[0],
        f'tvalue_target': results.tvalues.iloc[1],
        'conf_int_lowerbound_const': results.conf_int().iloc[0, 0],
        'conf_int_upperbound_const': results.conf_int().iloc[0, 1],
        f'conf_int_lowerbound_target': results.conf_int().iloc[1, 0],
        f'conf_int_upperbound_target': results.conf_int().iloc[1, 1],
        'test1.overall_signal': test1,
        'test2.realness': test2,
        'test3.explanatory_power': test3,
        'is_meaningful': test1 and test2 and test3,
    }
    return out


def ols_analysis(
        exog, # Independent variable,
        endogs, # Dependent variables
        sort='rsquared_adj'
        ):
    """
    Perform Ordinary Least Squares (OLS) analysis on the provided data.

    :param exog: The independent variable
    :param endogs: The dependent variables
    :poaram sort: The column to sort the results by.
    :type sort: str
    :return: DataFrame
    """

    # Dict to turn into a DataFrame
    result = {}

    # Add a column of 1s to the independent variable
    y = sm.add_constant(exog)

    for endog in endogs:
        # Create the model
        model = sm.OLS(endog, y)
        results = model.fit()

        log.debug(f'OLS for "{endog.name} == {exog.name}?"')
        log.debug(results.summary())

        # Check if OLS is meaningful
        meaningful_results = is_ols_meaningful(results)

        # Store the results
        result[endog.name] = meaningful_results
        result[endog.name]['model'] = results

    # Create the dataframe
    df_ols = pd.DataFrame(result).T
    df_ols.name = f'"{exog.name}": OLS'

    # The models use the same independent variable, sample sizes are the same,
    # and models are comparable so it is safe to sort by R²
    if sort:
        df_ols = df_ols.sort_values(by=sort, ascending=False)
    return df_ols

def get_meaningful_ols(df, exog=None, endogs=None, fillna=None):
    """

    :param df: DataFrame to test
    :param exog: Independent variable to test against
    :param exog: str
    :param endogs: Dependent variables to test for significance
    :type endogs: list
    :param fillna: dictionary of column names and aggregate method
    :type fillna: dict
    :return: DataFrame
    """

    # Check if we passed in endogs to specifically test against, otherwise use
    # all numeric columns in DataFrame
    endogs = endogs or df.columns.tolist()

    # Check fillna methods (some columns might want to use another aggregate func)
    if not fillna:
        fillna = {}
        for col in df:
            fillna[col] = 'mean'

    if not is_int_or_float(df[exog]):
        return

    # Remove the exog from the endogs list (because we're not testing
    # against itself)
    to_check = endogs.copy()
    if exog in to_check:
        to_check.remove(exog)

    # Fill null values with method
    check_endogs = []
    for check in to_check:
        if is_int_or_float(df[check]):
            check_endogs.append(df[check].fillna(getattr(df[check], fillna[check])()))

    # Run the analysis
    try:
        df_ols = ols_analysis(
            df[exog].fillna(getattr(df[exog], fillna[exog])()),
            check_endogs
        )
    except:
        pass

    df_ols_meaninginful = df_ols[df_ols['is_meaningful']]
    df_ols_meaninginful.name = exog

    # Store the RegressionResults who passed the meaningful test
    return df_ols_meaninginful
```

## OLS Viz

```python
def ols_viz(df,
            exog,
            endogs=None,
            fillna=None,
            height=5):
    """
    """


    df_ols = get_meaningful_ols(df, exog, endogs)

    # If no meaningful OLS found, raise ValueError
    if df_ols.empty:
        raise ValueError(f'No meaningful OLS found for "{exog}"')
        return

    # Create the figure and axes
    length = len(df_ols.index)
    if length == 1:
        fig, axes = plt.subplots(nrows=1, ncols=1, figsize=(20, height), layout='constrained')
        axes_flat = [axes]
    else:
        rows = int(length / 2)
        if length % 2 != 0:
            rows += 1
        fig, axes = plt.subplots(nrows=rows, ncols=2, figsize=(20, height*rows), layout='constrained')

        if rows >= 2:
            axes_flat = [item for sublist in axes for item in sublist]
        else:
            axes_flat = axes

    print(axes_flat, rows)
    # Set the figure title
    fig.suptitle(f'OLS Analysis for {exog}', **TITLE_KKWARGS)

    # Create the plots
    for i, (row, values) in enumerate(df_ols.iterrows()):

        exog = df[df_ols.name].fillna(df[df_ols.name].mean())
        endog = df[row].fillna(df[row].mean())
        model = values['model']

        # Create the scatter plot
        axes_flat[i].scatter(
            exog,
            endog,
            color='orange',
            alpha=0.6
        )

        # Create the regression line
        axes_flat[i].plot(
            exog,
            model.fittedvalues,
            color='red',
            label='Regression Line'
        )

        # Label some things
        title = f'{df_ols.name} -> {row}? | \
coeff={values.coeff_target: .2f} | \
f={values.fvalue: .2f} | \
p={values.pvalue_target: .2f} | \
rsq_adj={values.rsquared_adj: .2f} | \
t={values.tvalue_target: .2f}'
        axes_flat[i].set_title(title)
        axes_flat[i].set_ylabel(row, **SUB_TITLE_KKWARGS)

    plt.show()

# test = ols_viz(df_health, 'Calories', height=5)
```

## 💧 How Many Cups of Water Do I Drink a Day?

Water Intake tracks daily fluid consumption and is compared against a personalized hydration goal calculated by Garmin based on factors such as height, weight, and activity level. This metric provides context for daily hydration habits and helps assess consistency relative to recommended intake levels.

### Key Insights
* The recommended daily water intake is approximately 12 cups (96 oz), based on personalized hydration guidelines.
* Average daily water consumption is approximately 10.89 cups, indicating generally consistent hydration behavior.
